```python
!pip install -U google-genai
from google import genai

client = genai.Client(api_key="")


MODEL = "models/gemini-3.5-flash"


def health_agent(user_input):

    prompt = f"""
You are an AI healthcare assistant that provides general health guidance and triages
the urgency of a user's symptoms. You are NOT a doctor and must never give a definite
diagnosis or prescribe medication doses.

Provide a well-structured response with clear headings:

1. **Urgency Level**
   - Classify as: Critical (seek emergency care immediately), Urgent (see a doctor within 24 hours),
     or Normal (can be managed at home / routine checkup)
   - Briefly explain why

2. **Possible Causes**
   - List 2-4 general, common possible causes for these symptoms (not a diagnosis)

3. **Home Care Tips**
   - Practical, safe self-care steps the person can take right now

4. **When to See a Doctor (Red Flags)**
   - Specific warning signs that mean they should seek medical attention immediately

5. **Recommended Next Step**
   - Suggest whether to visit a nearby clinic, hospital ER, or just monitor at home

Keep the tone calm, clear, and easy to understand for a non-medical person.
Always end with this disclaimer on its own line:
"⚠️ This is general information only, not a medical diagnosis. Please consult a qualified doctor for proper care."

User's symptoms/request: {user_input}

Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text



print("Gemini Health Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = health_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
