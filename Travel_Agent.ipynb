```python
FLIGHTS = [
    {
        "name": "PIA PK-301",
        "source": "Karachi",
        "destination": "Lahore",
        "class": "Economy",
        "fare": 18000,
        "departure": "8:00 AM",
    },
    {
        "name": "Airblue PA-401",
        "source": "Karachi",
        "destination": "Lahore",
        "class": "Business",
        "fare": 32000,
        "departure": "11:30 AM",
    },
    {
        "name": "Fly Jinnah 9P-501",
        "source": "Karachi",
        "destination": "Islamabad",
        "class": "Economy",
        "fare": 20000,
        "departure": "2:00 PM",
    },
    {
        "name": "PIA PK-368",
        "source": "Karachi",
        "destination": "Islamabad",
        "class": "Business",
        "fare": 35000,
        "departure": "6:00 PM",
    },
    {
        "name": "Airblue PA-402",
        "source": "Lahore",
        "destination": "Karachi",
        "class": "Economy",
        "fare": 19000,
        "departure": "9:30 AM",
    },
    {
        "name": "PIA PK-305",
        "source": "Lahore",
        "destination": "Karachi",
        "class": "First",
        "fare": 40000,
        "departure": "3:00 PM",
    },
    {
        "name": "Fly Jinnah 9P-701",
        "source": "Islamabad",
        "destination": "Karachi",
        "class": "Economy",
        "fare": 21000,
        "departure": "10:00 AM",
    },
    {
        "name": "Airblue PA-207",
        "source": "Islamabad",
        "destination": "Lahore",
        "class": "Business",
        "fare": 15000,
        "departure": "7:30 AM",
    },
]


def get_user_input():
    """Ask for and normalize the travel details."""

    source = input("Enter source city: ").strip().title()
    destination = input("Enter destination city: ").strip().title()
    flight_class = input(
        "Enter class (Economy/Business/First): "
    ).strip().title()

    while True:
        budget_text = input("Enter your maximum budget (PKR): ").strip()

        try:
            budget = int(budget_text)

            if budget < 0:
                print("Budget cannot be negative. Please try again.")
            else:
                break

        except ValueError:
            print("Please enter the budget as a whole number.")

    return source, destination, flight_class, budget


def find_flight(source, destination, requested_class, budget):
    """Apply route, class, and budget rules."""

    route_found = False
    class_found = False
    over_budget_flight = None

    for flight in FLIGHTS:

        # Rule 1: Check route
        if (
            flight["source"] != source
            or flight["destination"] != destination
        ):
            continue

        route_found = True

        # Rule 2: Check class
        if flight["class"] != requested_class:
            continue

        class_found = True

        # Rule 3: Check budget
        if flight["fare"] <= budget:

            reasons = [
                "✓ Source and destination match the flight route",
                "✓ Flight class matches the requested class",
                "✓ Fare is within your budget",
            ]

            return flight, reasons, None

        # Flight matches route and class but is too expensive
        over_budget_flight = flight

    # No suitable flight was found
    if not route_found:
        hint = "Hint: no flight covers that route."

    elif not class_found:
        hint = "Hint: the route exists, but not in the requested class."

    else:
        hint = "Hint: matching flights were found, but they exceed your budget."

    return None, [], (over_budget_flight, hint)


def show_result(flight, reasons, failure):
    """Display the recommendation or failure message."""

    if flight is not None:

        print("\nRecommended Flight")
        print(f"Name: {flight['name']}")
        print(f"Departure: {flight['departure']}")
        print(f"Class: {flight['class']}")
        print(f"Fare: PKR {flight['fare']}")

        print("Why this flight matches:")

        for reason in reasons:
            print(reason)

        return

    over_budget_flight, hint = failure

    print("\nNo suitable flight was found")

    if over_budget_flight is not None:
        print(
            f"{over_budget_flight['name']} matches the route and class, "
            f"but its fare of PKR {over_budget_flight['fare']} "
            f"exceeds your budget."
        )

    print(hint)


def main():

    print("Pakistan Flight Rule-Based Agent")

    source, destination, requested_class, budget = get_user_input()

    flight, reasons, failure = find_flight(
        source,
        destination,
        requested_class,
        budget
    )

    show_result(flight, reasons, failure)


if __name__ == "__main__":
    main()
```
